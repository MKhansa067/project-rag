Setup & Import

```python
import os
from pathlib import Path

from google import genai
from google.genai import types
from IPython.display import Markdown
```

Load API key dari file `.env` dan buat client

```python
env_path = Path(".env")

if env_path.exists():
    for line in env_path.read_text().splitlines():
        line = line.strip()
        # lewati baris kosong, komentar, dan baris tanpa tanda "="
        if not line or line.startswith("#") or "=" not in line:
            continue
        key, value = line.split("=", 1)
        os.environ[key.strip()] = value.strip().strip('"').strip("'")

# Mengambil API key dari variabel lingkungan
api_key = os.getenv("GENAI_API_KEY") or os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError(
        "API key tidak ditemukan. Pastikan file .env berada di folder yang sama dengan notebook "
        "dan berisi baris: GENAI_API_KEY=isi_api_key_kamu"
    )

client = genai.Client(api_key=api_key)
```

```python
# Cek daftar model yang tersedia
for m in client.models.list():
    print(m.name)
```

Dengan base64(error)

```python
import base64, httpx

model = "gemini-3.1-flash-lite"
doc_url = "https://discovery.ucl.ac.uk/id/eprint/10089234/1/343019_3_art_0_py4t4l_convrt.pdf"  # Replace with the actual URL of your PDF

# Retrieve and encode the PDF
doc_data = base64.standard_b64decode(httpx.get(doc_url).content).decode("utf-8")

prompt = "Summarize this document"
response = client.models.generate_content(
    model=model,
    contents=[prompt,
              {
                  "inline_data": {
                      "mime_type": "application/pdf",
                      "data": doc_data
                  }
              }]
)
display(Markdown(response.text))
```

Dengan base64(no error)

```python
import base64, httpx

resp = httpx.get(doc_url, follow_redirects=True, timeout=60)
resp.raise_for_status()

doc_data = base64.standard_b64encode(resp.content).decode("utf-8")

response = client.models.generate_content(
    model=model,
    contents=[
        prompt,
        {"inline_data": {"mime_type": "application/pdf", "data": doc_data}},
    ],
)
display(Markdown(response.text))
```

Tanpa base64

```python
import httpx
from google.genai import types
from IPython.display import Markdown, display

model = "gemini-3.1-flash-lite"
doc_url = "https://discovery.ucl.ac.uk/id/eprint/10089234/1/343019_3_art_0_py4t4l_convrt.pdf"

# Ambil PDF (follow_redirects supaya redirect ikut diikuti, raise_for_status supaya error HTTP terlihat jelas)
resp = httpx.get(doc_url, follow_redirects=True, timeout=60)
resp.raise_for_status()
doc_bytes = resp.content

prompt = "Summarize this document"
response = client.models.generate_content(
    model=model,
    contents=[
        types.Part.from_bytes(data=doc_bytes, mime_type="application/pdf"),
        prompt,
    ],
)
display(Markdown(response.text))
```
